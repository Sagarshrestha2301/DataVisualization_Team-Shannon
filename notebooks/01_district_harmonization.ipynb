```python
districts = (
    census[
        (census["dist"] != 0) &
        (census["gapa"] == 0)
    ][["prov", "provname", "dist", "dname"]]
    .drop_duplicates()
    .sort_values(["prov", "dist"])
    .reset_index(drop=True)
)

districts.head()
```

```python
print("Rows:", len(districts))
print("Unique districts:", districts["dname"].nunique())
print("Unique district codes:", districts["dist"].nunique())
```

```python
print(districts["provname"].unique())
```

```python
province_lookup = {
    "Province-1": "Koshi",
    "PROVINCE - 1": "Koshi",
    "Madhesh": "Madhesh",
    "Bagmati": "Bagmati",
    "Gandaki": "Gandaki",
    "Lumbini": "Lumbini",
    "Karnali": "Karnali",
    "Sudurpaschim": "Sudurpashchim"
}

districts["province_std"] = districts["provname"].replace(province_lookup)
```

```python
districts["province_std"].unique()
```

```python
districts["district_std"] = districts["dname"].str.title()
```

```python
canonical_districts = districts[
    [
        "prov",
        "province_std",
        "dist",
        "district_std"
    ]
].rename(columns={
    "prov": "province_code",
    "province_std": "province_name",
    "dist": "district_code",
    "district_std": "district_name"
})

canonical_districts.head()
```

```python
canonical_districts.to_csv(
    "../data/lookup/canonical_districts.csv",
    index=False
)
```

```python
local_levels = (
    census[
        (census["gapa"] != 0) &
        (census["gapa"] != 99)
    ][
        [
            "prov",
            "dist",
            "gapa",
            "provname",
            "dname",
            "gapaname"
        ]
    ]
    .drop_duplicates()
    .sort_values(["prov","dist","gapa"])
    .reset_index(drop=True)
)
```

```python
print(len(local_levels))
```

```python
local_levels.duplicated(
    subset=["prov","dist","gapa"]
).sum()
```

```python
local_levels["province_name"] = local_levels["provname"].replace(province_lookup)

local_levels["district_name"] = local_levels["dname"].str.title()

local_levels["local_level_name"] = local_levels["gapaname"].str.title()

canonical_local_levels = (
    local_levels[
        [
            "prov",
            "province_name",
            "dist",
            "district_name",
            "gapa",
            "local_level_name"
        ]
    ]
    .rename(columns={
        "prov": "province_code",
        "dist": "district_code",
        "gapa": "local_level_code"
    })
)

canonical_local_levels.to_csv(
    "../data/lookup/canonical_local_levels.csv",
    index=False
)
```
